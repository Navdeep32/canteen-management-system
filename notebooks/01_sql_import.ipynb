```python
import pandas as pd
```

```python
import json
from sqlalchemy import create_engine

# Load secrets
with open('secrets.json', 'r') as f:
    secrets = json.load(f)

# Create connection
engine = create_engine(
    f"mysql+pymysql://{secrets['DB_USER']}:{secrets['DB_PASSWORD']}@{secrets['DB_HOST']}/{secrets['DB_NAME']}"
)

# Test connection
try:
    connection = engine.connect()
    print("Connected to database!")
    connection.close()
except Exception as e:
    print(f"Connection failed: {e}")
```

```python
sales_query = "SELECT * FROM sales"
inventory_query = "SELECT * FROM inventory"

sales_df = pd.read_sql(sales_query, engine)
inventory_df = pd.read_sql(inventory_query, engine)
```
